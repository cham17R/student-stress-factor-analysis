# Exploring Factors Associated with Student Stress

### A Behavioral Data Analysis Using Student Survey Responses

## Objective

This project explores factors that may be associated with stress among students, including optimism, sleep duration, social support, and work-life balance.

The analysis was conducted using survey responses collected from students and analyzed using Python, pandas, and Jupyter Notebook.

## Dataset Overview

The dataset consists of 20 student survey responses covering topics related to optimism, stress, sleep habits, social support, and lifestyle behaviors.

The survey was designed to explore possible relationships between emotional, social, and behavioral factors and student stress levels.

```python
import pandas as pd
```

```python
df = pd.read_csv("Optimism and Stress Level Study.csv")
```

```python
df.head()
```

```python
df.info()
```

## Data Cleaning

The dataset was cleaned by:
- Removing empty or unnecessary columns
- Renaming long column titles for easier analysis
- Checking for missing values and formatting consistency

```python
df= df.drop(columns=["@"])
```

```python
df.head()
```

# Optimism vs Stress

```python
df = df.rename(columns={
    "On a scale of 1 to 5, how optimistic do you generally feel about your future academic and career prospects? (1=Not at all optimistic, 5=Extremely optimistic)": "optimism_score",

    "On a scale of 1 to 5, how would you rate your current overall stress level as a student? (1=No stress, 5=Severe stress)": "stress_score"
})
```

```python
df[["optimism_score", "stress_score"]].head()
```

```python
df[["optimism_score", "stress_score"]].corr()
```

### Insight

The analysis showed little to no direct correlation between optimism and overall stress level among the surveyed students. This suggests that academic stress may be influenced by multiple factors beyond general optimism alone.

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df["optimism_score"], df["stress_score"])

plt.title("Optimism vs Stress Level")
plt.xlabel("Optimism Score")
plt.ylabel("Stress Score")

plt.show()
```

## Insight

The scatter plot shows little visible relationship between optimism scores and stress levels among the surveyed students. While some optimistic students reported lower stress, others still experienced moderate to high stress levels.

This suggests that academic stress may be influenced by multiple factors beyond optimism alone.

```python
df["stress_score"].value_counts()
```

```python
df["stress_score"].value_counts().sort_index().plot(kind="bar")

plt.title("Distribution of Student Stress Levels")
plt.xlabel("Stress Score")
plt.ylabel("Number of Students")

plt.show()
```

## Insight

The distribution shows that most students reported moderate to high stress levels, with stress scores of 3 and 4 being the most common. This suggests that academic stress is a common experience among the surveyed students.

# Sleep vs Stress

```python
df = df.rename(columns={
    "  On average, how many hours do you sleep each night?  ": "sleep_hours"
})
```

```python
df["sleep_hours"].value_counts()
```

```python
df.groupby("sleep_hours")["stress_score"].mean()
```

```python
df.groupby("sleep_hours")["stress_score"].mean().plot(kind="bar")

plt.title("Average Stress Level by Sleep Duration")
plt.xlabel("Sleep Duration")
plt.ylabel("Average Stress Score")

plt.show()
```

## Insight

Students who slept less than 5 hours per night reported the highest average stress levels. In contrast, students who slept 7–8 hours showed lower average stress levels.

This suggests that sleep duration may play an important role in student stress and well-being.

# Social Support vs Stress

```python
df = df.rename(columns={
    "  How supported do you feel by friends, classmates, or family during stressful times?  ": "support_level"
})
```

```python
df["support_level"].value_counts()
```

```python
df.groupby("support_level")["stress_score"].mean()
```

```python
df.groupby("support_level")["stress_score"].mean().plot(kind="bar")

plt.title("Average Stress Level by Support Level")
plt.xlabel("Support Level")
plt.ylabel("Average Stress Score")

plt.show()
```

## Insight

Students who reported lower levels of emotional or social support tended to experience higher stress levels. Participants with the lowest support levels reported the highest average stress scores.

This suggests that social support may play an important role in helping students manage academic stress.


# Work-Life Balance vs Stress

```python
df = df.rename(columns={
    "How would you describe your overall work-life balance as a student?  ": "work_life_balance"
})
```

```python
df["work_life_balance"].value_counts()
```

```python
df.groupby("work_life_balance")["stress_score"].mean()
```

## Insight

Students reporting a very poor work-life balance showed higher average stress levels. However, the relationship between work-life balance and stress was less consistent across the other categories.

This may suggest that extreme imbalance is associated with stress, while moderate differences are less clear within the current sample.
